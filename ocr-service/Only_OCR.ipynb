```python
import os
import glob
import time
from dotenv import load_dotenv
load_dotenv()
```

```python
from typhoon_ocr import ocr_document

input_folder = r"Enter the path to your image folder here"
output_folder = input_folder
image_files = glob.glob(os.path.join(input_folder, "*.[jJ][pP][gG]")) + \
              glob.glob(os.path.join(input_folder, "*.[pP][nN][gG]"))
print(f"พบไฟล์รูปภาพทั้งหมด {len(image_files)} ไฟล์")


# 4. เริ่ม Loop จัดการทีละไฟล์
for path_file in image_files:
    try:
        print(f"กำลังประมวลผล: {os.path.basename(path_file)}...")
        
        # ทำ OCR ด้วย Typhoon
        markdown = ocr_document(
            path_file, 
            model="typhoon-ocr", 
            figure_language="Thai", 
            task_type="v1.5"
        )
        
        # ดึงชื่อไฟล์ออกมาเพื่อตั้งชื่อ .txt
        base_name = os.path.basename(path_file)
        file_name_without_ext = os.path.splitext(base_name)[0]
        output_filename = os.path.join(output_folder, f"{file_name_without_ext}.txt")
        
        # บันทึกไฟล์
        with open(output_filename, "w", encoding="utf-8") as f:
            f.write(markdown)
            
        print(f"สำเร็จ! บันทึกไปที่: {output_filename}")
        time.sleep(4)

    except Exception as e:
        print(f"เกิดข้อผิดพลาดกับไฟล์ {path_file}: {e}")

print("--- ประมวลผลครบทุกไฟล์เรียบร้อยแล้ว ---")

```
